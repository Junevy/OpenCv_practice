# NumPy 核心练习：连接 Python、OpenCV 与 PyTorch/YOLO

这份练习不依赖 `numpy_100.ipynb`，按照“Python → NumPy → OpenCV → PyTorch → YOLO/深度学习”的路线设计。请你在每个空白代码单元中完成题目；Notebook 不提供答案。

练习重点是后续处理图像、批量数据和模型输入时最常用的数组概念与算子：`shape`、`ndim`、`dtype`、`axis`、广播、索引、布尔掩码、形状变换、view/copy、向量化、归约、随机数、图像布局转换和批量数据组织。

建议按顺序完成。除题目特别说明外，尽量使用 NumPy 向量化操作，避免对数组元素逐个写 Python `for` 循环。每道题后都要检查结果的形状、数据类型和数值范围。


## 一、数组模型与基本属性


### 01｜导入 NumPy 并确认环境

导入 NumPy，打印 NumPy 版本。创建一个简单的一维数组，确认它的类型是 `numpy.ndarray`。


### 02｜Python 列表与 ndarray

分别创建一个 Python 嵌套列表和一个 NumPy 二维数组。比较它们的类型、索引方式、逐元素加法结果，并说明为什么 NumPy 数组适合批量数值计算。


### 03｜理解 shape、ndim、size、dtype

创建形状为 `(2, 3, 4)` 的数组，输出 `shape`、`ndim`、`size`、`dtype`、`itemsize` 和 `nbytes`。用公式验证 `nbytes` 与其他属性的关系。


### 04｜理解 axis

创建一个 `3×4` 数组，分别沿 `axis=0` 和 `axis=1` 求和。画出或用文字说明每个 axis 表示的方向，并验证结果形状。


### 05｜dtype 与类型转换

创建 `int32`、`float32` 和 `uint8` 数组，比较它们的 `dtype`、`itemsize` 和取值范围。使用 `astype` 将数组转换为 `float32`，观察转换前后的结果。


## 二、数组创建与随机数据


### 06｜创建占位数组

分别使用 `zeros`、`ones`、`full` 和 `empty` 创建指定形状、指定 dtype 的数组。检查每个数组的形状和 dtype；不要假设 `empty` 的初始值。


### 07｜arange 与 linspace

使用 `arange` 生成整数序列，使用 `linspace` 生成等间距浮点序列。比较两者对终点、步长和元素个数的控制方式。


### 08｜单位矩阵、对角线与对角矩阵

使用 `eye` 或 `identity` 创建单位矩阵；提取一个二维数组的对角线；再用 `diag` 根据一维数组创建对角矩阵。检查每一步的形状。


### 09｜可复现的随机数

创建 `np.random.default_rng` 随机数生成器，设置固定种子，分别生成均匀分布、正态分布、随机整数和随机排列。重复运行并确认固定种子会得到相同结果。


### 10｜生成模拟数据集

生成 100 个样本、每个样本 4 个特征的 `float32` 数组，并生成对应的整数标签。检查数据和标签的形状、dtype，以及每个类别的样本数量。


## 三、索引、切片与条件筛选


### 11｜基本索引与切片

创建一个 `4×5` 数组，分别取单个元素、某一行、某一列、前两行、最后一列和步长为 2 的切片。为每个结果打印 shape。


### 12｜花式索引

使用整数索引一次选取多个指定行和多个指定列。分别比较连续切片与花式索引返回结果的形状。


### 13｜布尔掩码

创建一个包含正数、负数和零的数组，使用比较运算生成布尔掩码，筛选出大于 0 的元素以及绝对值大于某个阈值的元素。


### 14｜where 与 nonzero

使用 `where` 将数组中的负数替换为 0、非负数保持不变；使用 `nonzero` 获取满足条件的坐标，并用这些坐标回取原数组元素。


### 15｜clip、unique 与计数

创建一个包含重复值和异常范围值的数组。使用 `clip` 限制数值范围，使用 `unique(..., return_counts=True)` 统计每个不同值的出现次数。


## 四、向量化运算、归约与线性代数


### 16｜逐元素运算与逻辑运算

对两个同形状数组完成加、减、乘、除、取模、幂运算和比较运算。使用 `&`、`|`、`~` 组合多个布尔条件，并注意条件两侧需要加括号。


### 17｜聚合统计与 keepdims

对二维数组计算 `sum`、`mean`、`min`、`max`、`std`，分别指定 `axis=0`、`axis=1`，再使用 `keepdims=True` 比较结果形状。


### 18｜NaN、无穷值与安全统计

构造包含 `NaN` 和无穷值的数组，观察普通统计函数的结果；再使用 `isnan`、`isfinite` 和适当的 `nan*` 统计函数完成有效数据统计。


### 19｜累计运算与 reduce

对一维数组分别使用 `cumsum`、`cumprod`、`np.add.accumulate` 和 `np.add.reduce`。比较“保留每一步结果”和“只保留最终归约结果”的区别。


### 20｜矩阵乘法与 einsum

创建两个形状可相乘的二维数组，分别使用 `@`、`matmul` 和 `dot` 完成矩阵乘法。再用 `einsum` 表达同一个矩阵乘法，并检查结果是否一致。


## 五、形状变换、拼接与广播


### 21｜reshape 与 -1

创建 24 个连续元素，分别 reshape 成二维和三维数组。至少一次使用 `-1` 自动推断维度，并验证 reshape 前后元素总数不变。


### 22｜转置与轴交换

创建一个形状为 `(2, 3, 4)` 的数组，使用 `.T`、`transpose`、`swapaxes` 或 `moveaxis` 调整轴顺序，记录每一步的 shape。


### 23｜增加和删除长度为 1 的维度

使用 `expand_dims` 或 `None/newaxis` 增加行、列和批次维度；使用 `squeeze` 删除长度为 1 的维度。验证哪些维度可以安全删除。


### 24｜concatenate 与 stack

准备两个兼容数组，分别使用 `concatenate`、`vstack`、`hstack` 和 `stack` 拼接。重点比较 `concatenate` 不新增维度，而 `stack` 会新增维度。


### 25｜split 与 array_split

将一个一维数组和一个二维数组分别拆分为多个部分，比较 `split` 与 `array_split` 在不能平均分割时的行为。


### 26｜广播规则

构造形状为 `(3, 1)` 和 `(1, 4)` 的数组，利用广播得到 `(3, 4)` 结果。写出广播发生前后的形状，并再举一个不能广播的形状组合。


### 27｜broadcast_to、ravel 与 flatten

使用 `broadcast_to` 显式生成广播视图；比较 `ravel`、`flatten` 和 reshape 为一维后的结果，检查它们是否共享原数组内存。


## 六、内存、视图与原地操作


### 28｜view 与 copy

从二维数组切片得到一个子数组，修改子数组后观察原数组是否变化；再用 `copy` 创建独立副本并重复实验。用 `shares_memory` 验证结论。


### 29｜out 参数与原地运算

使用 `np.add`、`np.multiply` 或 `np.divide` 的 `out` 参数，把结果写入预先分配的数组。比较普通表达式和 `out` 写法对结果数组身份的影响。


### 30｜只读数组

创建数组并将 `flags.writeable` 设置为 `False`，尝试修改元素，观察异常类型。再恢复可写状态并完成修改。


### 31｜连续内存与 dtype 准备

创建一个数组，转置后检查 `flags.c_contiguous`；使用 `ascontiguousarray` 得到 C 连续数组，并比较 shape、dtype 和内存连续性。说明为什么这对 OpenCV/PyTorch 数据交换有帮助。


## 七、面向 OpenCV 的图像数组练习


### 32｜灰度图像的数值范围

创建一个模拟灰度图像，形状为 `(height, width)`，dtype 为 `uint8`。将它转换为 `float32` 并归一化到 `[0, 1]`，再转换回 `uint8`，检查最小值、最大值和 dtype。


### 33｜RGB 图像与通道访问

创建形状为 `(height, width, 3)` 的 RGB 图像。分别取 R、G、B 通道，交换 R/B 通道模拟 RGB 与 BGR 转换，并验证结果 shape 不变。


### 34｜裁剪、填充与掩码

创建一张模拟图像和一个矩形区域，使用切片完成裁剪，使用 `pad` 添加边界；再创建同形状的布尔掩码，仅保留矩形区域内的像素。


### 35｜批量图像布局转换

创建形状为 `(N, H, W, C)` 的 NHWC 图像批次，使用 `transpose` 转换为 `(N, C, H, W)` 的 NCHW 布局，再转换回来并验证数值一致。


### 36｜按通道归一化

创建一个 RGB 图像批次和长度为 3 的通道均值、标准差数组，利用广播完成逐通道标准化。检查每个通道的统计结果和广播后的形状。


### 37｜坐标网格与图像几何

使用 `meshgrid` 生成图像坐标网格，计算每个像素到图像中心的距离，并生成一个圆形布尔掩码。检查网格和掩码的 shape。


### 38｜从掩码获取目标框

创建一张二值目标掩码，使用 `where` 或 `nonzero` 获取前景像素坐标，计算其最小外接矩形 `(x_min, y_min, x_max, y_max)`。同时处理“掩码为空”的情况。


## 八、面向 PyTorch/YOLO 的数据准备


### 39｜标签转 one-hot

创建一维整数类别标签和类别总数，使用 NumPy 生成 one-hot 编码。验证每一行只有一个 1，且编码结果 shape 正确。


### 40｜可复现的数据集划分

生成样本索引，使用固定随机数生成器打乱索引，按比例拆分训练集和验证集。检查两部分没有重复索引，并保证重复运行结果一致。


### 41｜批量索引与 mini-batch

创建样本矩阵和标签数组，按 batch size 生成连续 mini-batch。每个 batch 同时取数据和标签，并检查最后一个 batch 的大小。


### 42｜特征矩阵与 flatten

创建批量图像数组，将每张图像展平为特征向量，得到形状 `(N, features)` 的特征矩阵。验证 `N` 不变、每个样本的元素总数正确。


### 43｜向量化计算检测框 IoU

给定两组格式为 `(x1, y1, x2, y2)` 的检测框，使用广播和向量化运算计算两两 IoU 矩阵。处理相交区域为空和框面积为零的情况。


### 44｜综合数据管线

生成一批模拟 RGB 图像和标签，完成以下流程：固定随机性、检查输入 dtype/shape、将像素归一化、将 NHWC 转为 NCHW、生成 mini-batch，并在每一步打印形状和数值范围。
