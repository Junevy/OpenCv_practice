## reshape

```python
import numpy as np

arr = np.arange(6)
arr

tArr = arr.reshape(3,2)
tArr

thArr = arr.reshape(1,3,2)
thArr
thArr.shape
```

```python
import numpy as np

arr = np.arange(6)
arr

tarr = np.reshape(arr, (1,6))
np.reshape(tarr, (tarr.size,))
```
