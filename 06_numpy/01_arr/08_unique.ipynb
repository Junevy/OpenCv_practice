```python
import numpy as np

ndarr = np.array([1,1,2,3,4,5,6,6,6,10])
np.unique(ndarr)

v, ls = np.unique(ndarr, return_index=True)
v, ls
```

```python
v,ls = np.unique(ndarr, return_counts=True)
v, ls
```
