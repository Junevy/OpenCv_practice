# Blur image


```python
from pathlib import Path

# Locate the img folder beside 01_start from common Jupyter working directories.
image_dirs = (Path.cwd() / '../img', Path.cwd() / 'img', Path.cwd() / '07_opencv/img')
img_dir = next((p.resolve() for p in image_dirs if (p / 'img1.jpg').is_file() and (p / 'img2.jpg').is_file()), None)
if img_dir is None:
    raise FileNotFoundError('Cannot locate 07_opencv/img. Run from the workspace, 07_opencv, or 01_start directory.')

import cv2
import numpy as np
import matplotlib.pyplot as plt

img1_path = str(img_dir / 'img1.jpg')
img1 = cv2.imread(img1_path, cv2.IMREAD_GRAYSCALE)
if img1 is None:
    raise FileNotFoundError(f'Cannot read image: {img1_path}')
print(f'img1 shape: {img1.shape}, dtype: {img1.dtype}')
```

```python
kernel_size = 11
mean_blur = cv2.blur(img1, (kernel_size, kernel_size))
gaussian_blur = cv2.GaussianBlur(img1, (kernel_size, kernel_size), 0)
median_blur = cv2.medianBlur(img1, 51)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for axis, title, image in zip(
    axes,
    ('Mean', 'Gaussian', 'Median'),
    (mean_blur, gaussian_blur, median_blur),
):
    axis.imshow(image, cmap='gray', vmin=0, vmax=255)
    axis.set_title(title)
    axis.axis('off')
plt.tight_layout()
plt.show()
```

```python
bilateral_blur = cv2.bilateralFilter(img1, 9, 50, 140)

plt.figure(figsize=(8, 5))
plt.imshow(bilateral_blur, cmap='gray', vmin=0, vmax=255)
plt.title('Bilateral')
plt.axis('off')
plt.show()
```
