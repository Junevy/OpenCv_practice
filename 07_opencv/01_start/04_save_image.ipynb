# Save image


```python
from pathlib import Path

# Locate the img folder beside 01_start from common Jupyter working directories.
image_dirs = (Path.cwd() / '../img', Path.cwd() / 'img', Path.cwd() / '07_opencv/img')
img_dir = next((p.resolve() for p in image_dirs if (p / 'img1.jpg').is_file() and (p / 'img2.jpg').is_file()), None)
if img_dir is None:
    raise FileNotFoundError('Cannot locate 07_opencv/img. Run from the workspace, 07_opencv, or 01_start directory.')

import cv2
import numpy as np
import matplotlib.pyplot as plt

img_path = str(img_dir / 'img1.jpg')
img = cv2.imread(img_path, cv2.IMREAD_COLOR)
if img is None:
    raise FileNotFoundError(f'Cannot read image: {img_path}')
print(f'Image shape: {img.shape}, dtype: {img.dtype}')
```

```python
from pathlib import Path

output_dir = Path.cwd() / 'opencv_demo_output'
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / 'saved_image.jpg'
if not cv2.imwrite(str(output_path), img):
    raise OSError(f'Could not save image to {output_path}')
saved_img = cv2.imread(str(output_path))
if saved_img is None or saved_img.shape != img.shape:
    raise OSError(f'Could not read back the saved image: {output_path}')
print(f'Saved image to: {output_path.resolve()}')
```
