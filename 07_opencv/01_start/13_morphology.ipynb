### Morphology

```python
from pathlib import Path

# Locate the img folder beside 01_start from common Jupyter working directories.
image_dirs = (Path.cwd() / '../img', Path.cwd() / 'img', Path.cwd() / '07_opencv/img')
img_dir = next((p.resolve() for p in image_dirs if (p / 'img1.jpg').is_file() and (p / 'img2.jpg').is_file()), None)
if img_dir is None:
    raise FileNotFoundError('Cannot locate 07_opencv/img. Run from the workspace, 07_opencv, or 01_start directory.')

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(str(img_dir / 'templ.png'))
img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(img, cmap='gray')

knl_size = np.ones((15,15), dtype=np.uint8)
```

```python
# eronsion
ero_img = cv2.erode(img, knl_size, 2)
plt.imshow(ero_img, 'gray')
```

```python
# dilation
# knl_size = np.ones((15,15), dtype=np.uint8)
dl_img = cv2.dilate(img, knl_size, 3)
plt.imshow(dl_img, 'gray')
```

```python
# opening
op_img = cv2.morphologyEx(img, cv2.MORPH_OPEN, knl_size)
plt.imshow(op_img, 'gray')
```

```python
# cloing
cl_img = cv2.morphologyEx(img, cv2.MORPH_CLOSE, knl_size)
plt.imshow(cl_img, 'gray')
```

```python
# test
sub_img = cv2.subtract(cl_img, img)
plt.imshow(sub_img, 'gray')
```

```python
# 形态学梯度
knl_size1 = np.ones((3,3), np.int8)
gr_img = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, knl_size1)
plt.imshow(gr_img, 'gray')
```
