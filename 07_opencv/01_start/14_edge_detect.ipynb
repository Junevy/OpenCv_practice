### edge detect

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('07_opencv\\img\\img2.jpg', cv2.IMREAD_GRAYSCALE)
knl_size = np.ones((5,5), dtype=np.uint8)
```

```python
# Canny
ceg_img = cv2.Canny(img, 50, 120)
er_img = cv2.dilate(ceg_img, knl_size, 2)
plt.imshow(er_img, 'gray')
```

```python
# sobel
sb_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sb_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sb_cmb = np.sqrt(sb_x**2 + sb_y**2)
# dl_img = cv2.dilate(sb_cmb, knl_size, 2)
plt.imshow(sb_x, 'gray')
```

```python
# Laplacian
lp_img = cv2.Laplacian(img, cv2.CV_64F)
# lp_img.max()
# lp_img.min()

# lp_abs_img = np.abs(lp_img)
# lp_abs_img.min()
# lp_abs_img.max()

lp_abs_img = cv2.convertScaleAbs(lp_img)
lp_abs_img.max()
lp_abs_img.mean()
lp_abs_img.shape
# plt.imshow(lp_abs_img, cmap='gray')
```
