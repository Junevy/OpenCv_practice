### import cv2

```python
from pathlib import Path

# Locate the img folder beside 01_start from common Jupyter working directories.
image_dirs = (Path.cwd() / '../img', Path.cwd() / 'img', Path.cwd() / '07_opencv/img')
img_dir = next((p.resolve() for p in image_dirs if (p / 'img1.jpg').is_file() and (p / 'img2.jpg').is_file()), None)
if img_dir is None:
    raise FileNotFoundError('Cannot locate 07_opencv/img. Run from the workspace, 07_opencv, or 01_start directory.')

import cv2
import numpy as np
import matplotlib.pyplot as plt

cv2.__version__
```

### read image

```python
img_path = str(img_dir / 'img1.jpg')
img = cv2.imread(img_path)

if img is None:
    print('read img error')
    exit()
type(img)
```

### show image

```python
# cv2.imshow('image', img)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis('off')
plt.show()
```

### split bgr

```python
b,g,r = cv2.split(img)
b,g,r
```

### merge b,g,r

```python
mg_img = cv2.merge([b,g,r])
plt.imshow(mg_img)
```

### resize & translation & rotate & flip

```python
# resized
resized_img = cv2.resize(img, (100, 100))
plt.imshow(resized_img)
```

```python
# rotate
h, w = img.shape[:2]
ctr = (w / 2, h / 2)
rtt_mtx = cv2.getRotationMatrix2D((ctr[0], ctr[1]), 30, 1.0)
rtt_img = cv2.warpAffine(img, rtt_mtx, (w, h))
plt.imshow(rtt_img)
```

```python
## translation
h, w = img.shape[:2]
tx, ty = w / 2, h / 2
trans_mtx = np.float32([[1,0,tx], [0,1,ty]])
trans_img = cv2.warpAffine(img, trans_mtx, (w, h))
plt.imshow(trans_img)
```

```python
# flip
flp_img = cv2.flip(img, 0)  # 0-vertical, 1-horizonal, 2-双向反转
plt.imshow(flp_img)
```

## imgage calculate

```python
img1 = cv2.imread(str(img_dir / 'img1.jpg'), cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread(str(img_dir / 'img2.jpg'), cv2.IMREAD_GRAYSCALE)
img1.shape
```

```python
# add
add_img = cv2.add(img1, img2)
plt.imshow(add_img, cmap='gray')
```

```python
# subtract
sub_img = cv2.subtract(img1, img2)
plt.imshow(sub_img)
```

```python
# 图像混合：叠加两张图像，计算公式： image1 * alpha + image2 * beta + garmma
fix_img = cv2.addWeighted(img2, 0.7, img1, 0.3, 0)
plt.imshow(fix_img)
```

### threshold

```python
# simple 
ret, thr_img = cv2.threshold(img1, 128, 255, cv2.THRESH_BINARY)
thr_img
print(ret)
plt.imshow(thr_img)
```

```python
# self_adaption threshold

print(img1.shape)
print(img1.dtype)

sel_img = cv2.adaptiveThreshold(img1, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 10)
plt.imshow(sel_img, cmap='gray')
```

```python
# Otus's threshold. 
# Otus's threshold is very fast but needs to keeping the image effective clear when use.
_, otus_img = cv2.threshold(img1, 0, 255, cv2.THRESH_OTSU)
plt.imshow(otus_img, cmap='gray')
```

## blurry

```python
knl_size = 11   ## kernel size
```

```python
# mean
mn_img = cv2.blur(img1, (knl_size, knl_size))
plt.imshow(mn_img, cmap='gray')
```

```python
# gaussian
gau_img = cv2.GaussianBlur(img1, (knl_size, knl_size), 0)   # Sigma: 决定权重的作用衰减（周围像素受权重影响的程度）
plt.imshow(gau_img, cmap='gray')
```

```python
# median
mk_size = 51
md_img = cv2.medianBlur(img1, mk_size)
plt.imshow(md_img, cmap='gray')
```

```python
# bilateral
bl_img = cv2.bilateralFilter(img1, mk_size, 50, 140)
plt.imshow(bl_img, cmap='gray')
```
