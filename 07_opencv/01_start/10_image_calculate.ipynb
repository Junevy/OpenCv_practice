# Image calculation


```python
from pathlib import Path

# Locate the img folder beside 01_start from common Jupyter working directories.
image_dirs = (Path.cwd() / '../img', Path.cwd() / 'img', Path.cwd() / '07_opencv/img')
img_dir = next((p.resolve() for p in image_dirs if (p / 'img1.jpg').is_file() and (p / 'img2.jpg').is_file()), None)
if img_dir is None:
    raise FileNotFoundError('Cannot locate 07_opencv/img. Run from the workspace, 07_opencv, or 01_start directory.')

import cv2
import numpy as np
import matplotlib.pyplot as plt

img1_path = str(img_dir / 'img1.jpg')
img1 = cv2.imread(img1_path, cv2.IMREAD_GRAYSCALE)
if img1 is None:
    raise FileNotFoundError(f'Cannot read image: {img1_path}')
print(f'img1 shape: {img1.shape}, dtype: {img1.dtype}')

img2_path = str(img_dir / 'img2.jpg')
img2 = cv2.imread(img2_path, cv2.IMREAD_GRAYSCALE)
if img2 is None:
    raise FileNotFoundError(f'Cannot read image: {img2_path}')
if img1.shape != img2.shape:
    raise ValueError(f'Image calculations require matching shapes: {img1.shape} != {img2.shape}')
print(f'img2 shape: {img2.shape}, dtype: {img2.dtype}')
```

```python
added_img = cv2.add(img1, img2)
subtracted_img = cv2.subtract(img1, img2)
weighted_img = cv2.addWeighted(img2, 0.7, img1, 0.3, 0)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for axis, title, image in zip(
    axes,
    ('Add', 'Subtract', 'Weighted blend'),
    (added_img, subtracted_img, weighted_img),
):
    axis.imshow(image, cmap='gray', vmin=0, vmax=255)
    axis.set_title(title)
    axis.axis('off')
plt.tight_layout()
plt.show()
```
